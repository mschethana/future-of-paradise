```python
COUNTRY_COL = "Pacific Island Countries and territories"
 
def clean_sdmx(df, value_name, keep_geo_pict=True):
    cols = {
        "GEO_PICT": "geo_pict",
        COUNTRY_COL: "country",
        "TIME_PERIOD": "year",
        "OBS_VALUE": value_name,
    }
    missing = [c for c in cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing expected columns: {missing}")
 
    out = df[list(cols.keys())].rename(columns=cols)
    out["year"] = out["year"].astype(int)
    out[value_name] = pd.to_numeric(out[value_name], errors="coerce")
    if not keep_geo_pict:
        out = out.drop(columns="geo_pict")
    return out
```

```python
sea_level   = clean_sdmx(read['sea_level'],   "sea_level_anomaly_m")
rainfall    = clean_sdmx(read['rainfall'],    "rainfall_anomaly_mm")
temperature = clean_sdmx(read['temp'], "temp_anomaly_c")
population  = clean_sdmx(read['population'],  "population_growth_pct")
tourism     = clean_sdmx(read['tourism'],     "tourist_arrivals")
affected    = clean_sdmx(read['affected_persons'], "persons_affected")
econ_loss   = clean_sdmx(read['econ_loss'],   "economic_loss_usd")
```

```python
temperature = temperature[temperature["year"] >= 1990].reset_index(drop=True)
```

```python
for name, df in [("tourism", tourism), ("econ_loss", econ_loss)]:
    countries = sorted(df["country"].unique())
    print(f"{name}: {countries}  ({'Tuvalu MISSING' if 'Tuvalu' not in countries else 'Tuvalu present'})")
```

```python
dfs = [sea_level, rainfall, temperature, population, tourism, affected, econ_loss]
 
master = dfs[0][["geo_pict", "country", "year"]].copy()
master = pd.concat([d[["geo_pict", "country", "year"]] for d in dfs]).drop_duplicates()
 
for df in dfs:
    value_col = [c for c in df.columns if c not in ("geo_pict", "country", "year")][0]
    master = master.merge(
        df[["country", "year", value_col]],
        on=["country", "year"],
        how="left"
    )
 
master = master.sort_values(["country", "year"]).reset_index(drop=True)
print(master.shape)
master.head(10)
```

```python
coverage = master.groupby("country").agg(
    {c: lambda x: x.notna().sum() for c in
     ["sea_level_anomaly_m", "rainfall_anomaly_mm", "temp_anomaly_c",
      "population_growth_pct", "tourist_arrivals", "persons_affected",
      "economic_loss_usd"]}
)
coverage
```

```python
os.makedirs("../data/processed", exist_ok=True)
master.to_csv("../data/processed/master_climate_data.csv", index=False)
print("Saved:", master.shape)
```
